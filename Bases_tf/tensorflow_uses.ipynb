```python
X = tf.constant([[1, 2],
                 [3, 4],
                 [5, 6]])

Y = tf.constant([[7, 8],
                 [9, 10],
                 [11, 12]])
Y.ndim
```

```python
X*Y
```

```python
# reshape des matrices
X @ tf.reshape(Y, shape=(2,3))
```

```python
# Transposer
tf.transpose(X)
```

```python
# multiplication des matrices (autres manière de faire la multiplication de matrices)
tf.matmul(tf.transpose(X), Y)
```

```python
# Autres formule
tf.matmul(a=X,b=Y,transpose_a=True, transpose_b=False)
```

```python
import numpy as np
E = tf.constant(np.random.randint(low = 0, high = 100, size=50))
E
```

```python
# trouver le minimum de mon tenseur
tf.reduce_min(E)
```

```python
# trouver le maximum
tf.reduce_max(E)
```

```python
# trouver la somme
tf.reduce_sum(E)
```

```python
#Utilisation de Argmax (trouver l'index de la valeur maximale dans le tensuer)

tf.argmax(E)
```

```python
#Utilisation de Argmin
tf.argmin(E)
```

```python
#SQUEEZE (reduire des dimensions)
G = tf.constant(np.random.randint(0, 100, 50), shape=(1, 1, 1, 1,50))
G.shape, G.ndim
```

```python
# enlever les dimensions de type 1 (par defaut, squeeze retire toutes les dimensions 1)
G_squeezed = tf.squeeze(G)
G_squeezed.shape, G_squeezed.ndim
```

```python
# one hot encoding
liste = [0, 1, 2, 3]
# encoder les valeurs de mon tenseur, la valeur de depth depends du nombre d'élements dans le tenseur
tf.one_hot(liste, depth=4)
```
